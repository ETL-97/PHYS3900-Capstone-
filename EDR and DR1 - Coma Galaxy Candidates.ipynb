```python
from astropy.table import Table
import pandas as pd
from matplotlib import pyplot as plt
import numpy as np
from astropy.coordinates import SkyCoord
import astropy.units as u

# convert .fits file -> table -> PANDAS

# Used Claude to determine how to convert to readable format 
full_data_dr1 = Table.read('FP_clustering_data_v5_combinedzp.fits')
for col in full_data_dr1.colnames:
    if full_data_dr1[col].dtype.byteorder == '>':
        full_data_dr1[col] = full_data_dr1[col].byteswap().newbyteorder()
full_data_dr1 = full_data_dr1.to_pandas()
full_data_edr = pd.read_csv('fuji_pv_full.csv', low_memory=False)

# line of sight code section ~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~


# DR1 ~~~~~~~~~~~~~~~~~~~~~

# identify Coma cluster candidate galaxies

# position in (Said et al., 2025).
coma_centre = SkyCoord('12h59m48.7s', '+27d58m50s', distance=99.1*u.Mpc, frame='icrs')

# define ra and dec of catalogue (in degrees)
dr1_coords = SkyCoord(ra=full_data_dr1['RA'].to_numpy()*u.deg, dec=full_data_dr1['DEC'].to_numpy()*u.deg, frame='icrs')

# compute separation 
sep_dr1 = coma_centre.separation(dr1_coords)

# define line of sight radius to Coma using separation parameter (2 deg. span (~3 deg diameter))
sep_param = 2*u.deg
los_dr1 = sep_param >= sep_dr1
candidate_gals_dr1 = full_data_dr1[los_dr1]

print('There are',len(candidate_gals_dr1),'galaxies within',sep_param,
      'of the centre of Coma in the DR1 sample.')
```

```python
# EDR ~~~~~~~~~~~~~~~~~~~~~~~

# define ra and dec of catalogue (in degrees)
edr_coords = SkyCoord(ra=full_data_edr['ra_1'].to_numpy()*u.deg, dec=full_data_edr['dec_1'].to_numpy()*u.deg, frame='icrs')

# compute separation 
sep_edr = coma_centre.separation(edr_coords)

# define line of sight radius to Coma using separation parameter (2 deg. span (~3 deg diameter))
los_edr = sep_param >= sep_edr
candidate_gals_edr = full_data_edr[los_edr]

print('There are',len(candidate_gals_edr),'galaxies within',sep_param,
      'of the centre of Coma in the EDR sample.')
```
